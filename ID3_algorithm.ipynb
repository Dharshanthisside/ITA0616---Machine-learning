```python

import pandas as pd
from sklearn.tree import DecisionTreeClassifier, export_text
from sklearn.preprocessing import LabelEncoder

# Dataset
data = {
    'Outlook': ['Sunny', 'Sunny', 'Rain', 'Rain', 'Overcast', 'Overcast'],
    'Humidity': ['High', 'Normal', 'High', 'Normal', 'High', 'Normal'],
    'Wind': ['Weak', 'Strong', 'Strong', 'Weak', 'Weak', 'Strong'],
    'Play': ['No', 'Yes', 'No', 'Yes', 'Yes', 'Yes']
}

df = pd.DataFrame(data)

# Convert text into numbers
encoders = {}
for col in df.columns:
    encoders[col] = LabelEncoder()
    df[col] = encoders[col].fit_transform(df[col])

# Build ID3 decision tree
X = df[['Outlook', 'Humidity', 'Wind']]
y = df['Play']

model = DecisionTreeClassifier(criterion='entropy', random_state=0)
model.fit(X, y)

# Display decision tree
print(export_text(model, feature_names=list(X.columns)))

# Classify a new sample
sample = pd.DataFrame([{
    'Outlook': 'Sunny',
    'Humidity': 'Normal',
    'Wind': 'Weak'
}])

for col in sample.columns:
    sample[col] = encoders[col].transform(sample[col])

prediction = model.predict(sample)[0]
print("Prediction:", encoders['Play'].inverse_transform([prediction])[0])
```
