```python

import numpy as np

# XOR dataset
X = np.array([[0,0], [0,1], [1,0], [1,1]])
y = np.array([[0], [1], [1], [0]])

def sigmoid(x):
    return 1 / (1 + np.exp(-x))

np.random.seed(1)
W1 = np.random.randn(2, 4) * 0.5
W2 = np.random.randn(4, 1) * 0.5
b1 = np.zeros((1, 4))
b2 = np.zeros((1, 1))

lr = 0.5

# Training using backpropagation
for epoch in range(10000):
    h = sigmoid(X @ W1 + b1)
    out = sigmoid(h @ W2 + b2)

    error = y - out
    d_out = error * out * (1 - out)
    d_h = (d_out @ W2.T) * h * (1 - h)

    W2 += lr * h.T @ d_out
    b2 += lr * d_out.sum(axis=0, keepdims=True)
    W1 += lr * X.T @ d_h
    b1 += lr * d_h.sum(axis=0, keepdims=True)

# Testing
print("Input -> Predicted output")
for x in X:
    h = sigmoid(x.reshape(1, 2) @ W1 + b1)
    out = sigmoid(h @ W2 + b2)
    print(x, "->", int(out[0, 0] >= 0.5))
```
