```python

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import train_test_split
from sklearn.naive_bayes import GaussianNB
from sklearn.metrics import accuracy_score, confusion_matrix

# Create dataset
data = {
    'Study_Hours': [1, 2, 3, 4, 5, 6, 7, 8, 2, 3,
                    5, 6, 7, 8, 4, 9, 1, 6, 8, 5],
    'Attendance': [40, 45, 50, 60, 65, 70, 80, 85, 90, 75,
                   55, 75, 85, 90, 65, 95, 35, 80, 95, 70],
    'Result': ['Fail', 'Fail', 'Fail', 'Pass', 'Pass',
               'Pass', 'Pass', 'Pass', 'Pass', 'Pass',
               'Fail', 'Pass', 'Pass', 'Pass', 'Pass',
               'Pass', 'Fail', 'Pass', 'Pass', 'Pass']
}

df = pd.DataFrame(data)

# Separate input and output
X = df[['Study_Hours', 'Attendance']]
y = df['Result']

# Split dataset
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=42
)

# Create Naive Bayes model
model = GaussianNB()

# Train the model
model.fit(X_train, y_train)

# Predict results
y_pred = model.predict(X_test)

# Display actual and predicted results
print("Actual Results:", list(y_test))
print("Predicted Results:", list(y_pred))

# Calculate accuracy
accuracy = accuracy_score(y_test, y_pred)
print("Accuracy:", accuracy * 100, "%")

# Generate confusion matrix
cm = confusion_matrix(y_test, y_pred, labels=['Fail', 'Pass'])

print("Confusion Matrix:")
print(cm)

# Display confusion matrix
sns.heatmap(
    cm, annot=True, fmt='d', cmap='Blues',
    xticklabels=['Fail', 'Pass'],
    yticklabels=['Fail', 'Pass']
)

plt.xlabel("Predicted Result")
plt.ylabel("Actual Result")
plt.title("Naive Bayes Confusion Matrix")
plt.show()

```
